  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">1. IMPORT LIBRARY</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat pustaka komputasi numerik, manipulasi data, visualisasi grafik, dan kerangka kerja PyTorch.</p>
</div>

```python
#202432068_Dhovinka Raezya Intan Febria
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import TensorDataset, DataLoader
from sklearn.preprocessing import MinMaxScaler
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">2. PRA-PEMROSESAN DATA (PREPROCESSING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Memuat dataset, membersihkan data, melakukan normalisasi / tokenisasi sekuens, serta membagi data latih dan data uji.</p>
</div>

```python
#202432068_Dhovinka Raezya Intan Febria

df = pd.read_csv('/cleaned_weather.csv')

print("Ukuran dataset:", df.shape)
print(df.head())

# 2. Membersihkan dan menyiapkan data
df['date'] = pd.to_datetime(df['date'])
df = df.sort_values('date')
df = df.dropna(subset=['T'])

data = df[['T']].values.astype(float)

# 3. Normalisasi data
scaler = MinMaxScaler()
jumlah_data = len(data)
batas_train = int(jumlah_data * 0.70)

scaler.fit(data[:batas_train])
data_scaled = scaler.transform(data)

# 4. Membuat urutan data (sequence)
window_size = 24
X, y = [], []

for i in range(len(data_scaled) - window_size):
    X.append(data_scaled[i:i + window_size])
    y.append(data_scaled[i + window_size])

X = np.array(X)
y = np.array(y)

# 5. Membagi data latih, validasi, dan uji
batas_train = int(len(X) * 0.70)
batas_val = int(len(X) * 0.85)

X_train, y_train = X[:batas_train], y[:batas_train]
X_val, y_val = X[batas_train:batas_val], y[batas_train:batas_val]
X_test, y_test = X[batas_val:], y[batas_val:]

# 6. Mengubah data menjadi Tensor PyTorch
X_train = torch.tensor(X_train, dtype=torch.float32)
y_train = torch.tensor(y_train, dtype=torch.float32)
X_val = torch.tensor(X_val, dtype=torch.float32)
y_val = torch.tensor(y_val, dtype=torch.float32)
X_test = torch.tensor(X_test, dtype=torch.float32)
y_test = torch.tensor(y_test, dtype=torch.float32)

print("Data preprocessing berhasil!")
print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("X_test:", X_test.shape)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">3. PEMODELAN ARSITEKTUR JARINGAN (RNN / LSTM)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Mendefinisikan kelas arsitektur model sekuensial menggunakan modul torch.nn sesuai algoritma yang dipilih.</p>
</div>

```python
#202432068_Dhovinka Raezya Intan Febria

class SequenceModel(nn.Module):
    def __init__(self):
        super(SequenceModel, self).__init__()

        # Layer LSTM
        self.lstm = nn.LSTM(
            input_size=1,
            hidden_size=64,
            num_layers=2,
            batch_first=True,
            dropout=0.2
        )

        # Dropout dan layer output
        self.dropout = nn.Dropout(0.2)
        self.fc = nn.Linear(64, 1)

    def forward(self, x):
        # Memproses urutan data menggunakan LSTM
        output, (hidden, cell) = self.lstm(x)

        # Mengambil output pada urutan terakhir
        output = output[:, -1, :]

        # Mengurangi risiko overfitting
        output = self.dropout(output)

        # Menghasilkan prediksi suhu
        output = self.fc(output)

        return output


# Membuat model LSTM
model = SequenceModel()

print(model)
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">4. INISIASI HYPERPARAMETER & PROSES PELATIHAN (TRAINING)</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menentukan konfigurasi hyperparameter (learning rate, epochs, loss function, optimizer) dan menjalankan siklus pelatihan model.</p>
</div>

```python
#202432068_Dhovinka Raezya Intan Febria

# 1. Inisialisasi hyperparameter
epochs = 30
learning_rate = 0.001
batch_size = 64

# 2. Inisialisasi model, fungsi loss, dan optimizer
model = SequenceModel()

criterion = nn.MSELoss()
optimizer = optim.Adam(model.parameters(), lr=learning_rate)

# 3. Menyimpan riwayat nilai loss
train_losses = []
val_losses = []

# 4. Proses pelatihan model
for epoch in range(epochs):

    model.train()
    optimizer.zero_grad()

    # Prediksi data training
    output_train = model(X_train)
    loss_train = criterion(output_train, y_train)

    # Menghitung gradien dan memperbarui bobot
    loss_train.backward()
    optimizer.step()

    # Evaluasi menggunakan data validasi
    model.eval()

    with torch.no_grad():
        output_val = model(X_val)
        loss_val = criterion(output_val, y_val)

    train_losses.append(loss_train.item())
    val_losses.append(loss_val.item())

    print(
        f"Epoch [{epoch+1}/{epochs}], "
        f"Train Loss: {loss_train.item():.6f}, "
        f"Validation Loss: {loss_val.item():.6f}"
    )

print("\nPelatihan model selesai!")
```

<div style="background-color: #114F6A; color: #FFFFFF; padding: 10px 18px; border-radius: 8px; border-left: 8px solid #1FA4B1; font-family: 'Segoe UI', sans-serif; margin-top: 16px;">
  <h3 style="margin: 0; font-size: 16px; color: #FFFFFF;">5. VISUALISASI LOSS TRAINING & PERBANDINGAN PREDIKSI VS AKTUAL</h3>
  <p style="margin: 4px 0 0 0; font-size: 12px; color: #E0F7FA;">Menampilkan grafik konvergensi nilai Loss terhadap Epoch dan grafik evaluasi perbandingan hasil prediksi model terhadap data aktual.</p>
</div>

```python
#202432068_Dhovinka Raezya Intan Febria

# 1. Grafik Loss Training dan Validasi
plt.figure(figsize=(10, 4))

plt.plot(train_losses, label='Training Loss', linewidth=2)
plt.plot(val_losses, label='Validation Loss', linewidth=2)

plt.title('Grafik Penurunan Loss Selama Pelatihan')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.show()


# 2. Prediksi Data Testing
model.eval()

with torch.no_grad():
    prediksi = model(X_test).numpy()

# Mengembalikan nilai ke skala suhu asli
prediksi_asli = scaler.inverse_transform(prediksi)
aktual_asli = scaler.inverse_transform(y_test.numpy())


# 3. Menghitung Evaluasi Model
from sklearn.metrics import mean_absolute_error, mean_squared_error

mae = mean_absolute_error(aktual_asli, prediksi_asli)
mse = mean_squared_error(aktual_asli, prediksi_asli)
rmse = np.sqrt(mse)

print("HASIL EVALUASI MODEL LSTM")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)


# 4. Grafik Perbandingan Prediksi dan Aktual
plt.figure(figsize=(12, 5))

plt.plot(aktual_asli[:200], label='Suhu Aktual')
plt.plot(prediksi_asli[:200], label='Suhu Prediksi')

plt.title('Perbandingan Suhu Aktual dan Prediksi LSTM')
plt.xlabel('Urutan Waktu')
plt.ylabel('Suhu Udara')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```
